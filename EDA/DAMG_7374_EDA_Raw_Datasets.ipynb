Loading CSV Files

```python
from pandas.errors import ParserError

# Paths to all 4 files
file_paths = [
    '/content/drive/MyDrive/Datasets Raw/raw_bloodpressure_data.csv',
    '/content/drive/MyDrive/Datasets Raw/raw_diabetes_data.csv',
    '/content/drive/MyDrive/Datasets Raw/raw_maternalhealth_data.csv',
    '/content/drive/MyDrive/Datasets Raw/raw_mentalhealth.csv',
    '/content/drive/MyDrive/Datasets Raw/raw_pcos_data.csv',
    '/content/drive/MyDrive/Datasets Raw/raw_pcos_infertility_data.csv',
      '/content/drive/MyDrive/Datasets Raw/raw_menstrual_data.csv',
]

# Load them into separate DataFrames
df_bp = pd.read_csv(file_paths[0])
df_diabetes = pd.read_csv(file_paths[1])
df_maternal= pd.read_csv(file_paths[2])
df_mental = pd.read_csv(file_paths[3])
df_pcos = pd.read_csv(file_paths[4])
df_infertility = pd.read_csv(file_paths[5])
df_menstrual= pd.read_csv(file_paths[6])


# Quick peek at each
print("Blood Pressure:")
display(df_bp.head())

print("Diabetes:")
display(df_diabetes.head())

print("Maternal:")
display(df_maternal.head())

print("Mental Health:")
display(df_mental.head())

print("Pcos:")
display(df_pcos.head())

print("Infertility:")
display(df_infertility.head())

print("Menstrual:")
display(df_menstrual.head())
```

1) Basic EDA of BP dataset

1a)  Checking General Info

```python
# Shape of the dataset
print("Shape:", df_bp.shape)

# Column names and data types
print("\nInfo:")
df_bp.info()

# First 5 rows
print("\nFirst 5 rows:")
display(df_bp.head())
```

1b) Checking for missing values

```python
# Count missing values per column
print("Missing values per column:")
df_bp.isnull().sum()
```

1c) Checking for summary info

```python
# Summary stats for numerical columns
print("Summary statistics:")
df_bp.describe()
```

1d) Checking for categorical columns

```python
# List of object (categorical) columns
categorical_cols = df_bp.select_dtypes(include='object').columns
print("Categorical columns:", categorical_cols)

# Show unique values for each categorical column
for col in categorical_cols:
    print(f"\nColumn: {col}")
    print(df_bp[col].value_counts())
```

1e) Distribution of numeric columns

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: make plots look nicer
sns.set(style="whitegrid")

numeric_cols = df_bp.select_dtypes(include='number').columns

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df_bp[col], kde=True, bins=30, color='skyblue')
    plt.title(f'Distribution of {col}')
    plt.show()
```

1f) Detecting outliers

```python
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(y=df_bp[col], color='lightgreen')
    plt.title(f'Boxplot of {col}')
    plt.show()
```

1g) Relationship between numerical values

```python
# Pairplot to see relationships
sns.pairplot(df_bp[numeric_cols])
plt.show()
```

1h) Counterplots for categorical columns

```python
categorical_cols = df_bp.select_dtypes(include='object').columns

for col in categorical_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(y=df_bp[col], order=df_bp[col].value_counts().index, palette='pastel')
    plt.title(f'Count of {col}')
    plt.show()
```

1i) Correlation heatmap

```python
plt.figure(figsize=(16,10))
sns.heatmap(df_bp[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
